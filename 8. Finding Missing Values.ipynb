### 3. Finding Missing Values
##### Task 2: Identify missing values for all columns.


```python
## Write your code here
missing_data = df.isnull().sum().sort_values(ascending=False)
missing_data
```

##### Task 3: Visualize missing values using a heatmap (Using seaborn library).



```python
import matplotlib.pyplot as plt
import seaborn as sns

missing_cols = df.columns[df.isnull().sum() > 0]
sample_df = df[missing_cols].sample(n=min(50, len(df)), random_state=42)

plt.figure(figsize=(10, 4))
sns.heatmap(sample_df.isna().T, cbar=False, cmap=["#2b2b2b", "#ffd700"])
plt.show()
```

##### Task 4: Count the number of missing rows for a specific column (e.g., `Employment`).


```python
# Count missing values in the Employment column
missing_data['Employment'].sum()
```

### 4. Imputing Missing Values
##### Task 5: Identify the most frequent (majority) value in a specific column (e.g., `Employment`).


```python
most_frequent_employment = df['Employment'].mode()[0]
most_frequent_employment
```

##### Task 6: Impute missing values in the `Employment` column with the most frequent value.



```python
# Fill missing values in Employment with that most frequent value
df['Employment'] = df['Employment'].fillna(most_frequent_employment)
```

### 5. Visualizing Imputed Data
##### Task 7: Visualize the distribution of a column after imputation (e.g., `Employment`).


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Порахувати частоту кожної категорії Employment
employment_counts = df['Employment'].value_counts()

# Залишити тільки категорії, які зустрічаються часто
threshold = 100
employment_counts = employment_counts[employment_counts >= threshold]

# Побудувати графік
plt.figure(figsize=(12, 6))
sns.barplot(
    x=employment_counts.values,
    y=employment_counts.index,
    palette='viridis'
)

plt.title(f'Employment Distribution After Imputation (count >= {threshold})')
plt.xlabel('Count')
plt.ylabel('Employment')
plt.tight_layout()
plt.show()
```
