### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset Into a Dataframe


#### **Read Data**
<p>
The functions below will download the dataset into your browser:
</p>


```python
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3. Finding and Removing Duplicates
##### Task 1: Identify duplicate rows in the dataset.


```python
df[df.duplicated()]
```

##### Task 2: Remove the duplicate rows from the dataframe.



```python
df = df.drop_duplicates()
df[df.duplicated()]
```

### Step 4: Finding Missing Values
##### Task 3: Find the missing values for all columns.


```python
missing_values=df.isnull()
missing_values
```

##### Task 4: Find out how many rows are missing in the column RemoteWork.


```python
df['RemoteWork'].isna().sum()
```

### Step 5. Imputing Missing Values
##### Task 5: Find the value counts for the column RemoteWork.


```python
df['RemoteWork'].value_counts()
```

##### Task 6: Identify the most frequent (majority) value in the RemoteWork column.



```python
majority_remote = df['RemoteWork'].mode()[0]
```

##### Task 7: Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
df['RemoteWork'] = df['RemoteWork'].fillna(majority_remote)
```

##### Task 8: Check for any compensation-related columns and describe their distribution.



```python
# Find compensation-related columns
comp_cols = [col for col in df.columns if 'Comp' in col or 'Salary' in col]
comp_cols
```

```python
# Describe the main compensation column
df['ConvertedCompYearly'].describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))
sns.histplot(df['ConvertedCompYearly'].dropna(), bins=50, kde=True, color='steelblue')
plt.title('Distribution of ConvertedCompYearly')
plt.xlabel('ConvertedCompYearly')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```
